# Pyplot API

```python
# Univariate - Numerical
```

```python
data = {
    "Salary" : [25000, 30000, 35000, 40000, 45000, 50000, 55000, 48957, 99563, 56387]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.shape
```

```python
# Line Plot

plt.plot(df["Salary"], color = "blue", marker = "o", linestyle = "--", linewidth = 3)
plt.show()

# We can change color, marker, linestye & linewidth according to our need
```

```python
# Histogram : Tells the frequency of the data

plt.hist(df["Salary"], bins = 4, color = "red")
plt.show()
```

```python
# Box PLot : 
plt.boxplot(df["Salary"])
plt.show()
```
