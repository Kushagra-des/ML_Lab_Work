# Scattering the Plot Using Different Parameters

```python
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g")
```

```python
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
```

```python
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "sex")
```

# Different Ways of Styling the Grid

```python
sns.set_style("ticks") # White, Dark, Whitegrid, Blackgrid, Ticks
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
```

```python
sns.set_style("dark")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
```

```python
sns.set_style("white")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
```

```python
sns.set_style("darkgrid")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
```

```python
sns.set_style("ticks")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
sns.despine(left = True)
```

```python
sns.set_style("ticks")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
sns.despine(right = True)
```

# Set_Context
Set_Context(): Sets the parameters that control the scaling of plot elements like label sizes, line thicknesses, and data point markers.

Types: Poster, Paper, Notebook, Talk

```python
sns.set_style("ticks")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island")
sns.set_context("notebook")
```

```python
sns.scatterplot(data = penguins, x = "species", y = "body_mass_g", hue = "island", style = "sex")
```
