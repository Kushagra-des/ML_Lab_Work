```python
data7 = pd.Series(['01 Jan 2015','10-02-2016','20180307','2014/05/06','2016-04-12','2019-04-06T11:20'])
print("Original Series : ")
print(data7)
print("")
data7 = data7.replace({'10-02-2016': '02-10-2016'})
data77 = pd.to_datetime(data7, format = 'mixed', dayfirst = True)
print("Day of month:")
print(list(data77.dt.day))
print("")
print("Day of year:")
print(list(data77.dt.dayofyear))
print("")
print("Week number:")
print(list(data77.dt.isocalendar().week.astype(int)))
print("")
print("Day of week:")
print(list(data77.dt.day_name()))
```

```python
exam_data = {'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin',
'Jonas'],'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

print("First three rows of the data frame : ")
data9 = pd.DataFrame(exam_data, index = labels)
data9 = data9[['attempts', 'name', 'qualify', 'score']]
print(data9.head(3))
```

```python
data10 = data9[data9['attempts']>2]
data10 = data10[['name', 'score', 'attempts', 'qualify']]
print(data10)
```

```python
data11 = data9[(data9['attempts']<2) & (data9['score']>15)]
print(data11)
```

```python
data12 = data9['attempts'].sum()
print(12)
```

```python
data13 = data9.sort_values(by=['name', 'score'], ascending=[False, True])
print(data13)
```
