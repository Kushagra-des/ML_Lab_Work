# BAR PLOT

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = ["Physics", "Chemistry", "Maths", "Biology", "English"]
y = [78,49,68,70,80]
```

```python
plt.xlabel("Subjects", fontsize = 10)
plt.ylabel("Marks", fontsize = 10)
plt.title("Report Data", fontsize = 10)
c = ["blue", "black", "red", "green", "magenta"]
plt.bar(x,y, width = 0.5, color = c)
plt.show()
```

```python
plt.bar(x,y, width = 0.5, color = c, align = "edge") # Align the text to edge
plt.show()
```

```python
plt.bar(x,y, width = 0.5, color = c, edgecolor = "yellow", linewidth = 3, linestyle = "--")
plt.show()
```

```python
plt.bar(x,y, width = 0.5, color = c, edgecolor = "yellow", linewidth = 3, alpha = 0.5)
plt.show()
```

```python
plt.bar(x,y, width = 0.5, color = c, edgecolor = "yellow", label = "1st Semester")
plt.legend()
plt.show()
```

```python
# Plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 2, 1) # (Rows, Columns, Number of graph)
plt.plot(x, y)

# Plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 2, 2)
plt.plot(x, y)

# Plot 3:
x = np.array([0, 1, 2, 5])
y = np.array([10, 20, 30, 50])

plt.subplot(2, 2, 3)
plt.plot(x, y)

# Plot 4:
x = np.array([0, 1, 2, 6])
y = np.array([10, 20, 30, 60])

plt.subplot(2, 2, 4)
plt.plot(x, y)

plt.suptitle("Subplot Title")
plt.show()
```
