```python
subjects = ['Math', 'Physics', 'Chemistry', 'English', 'CS']
marks = [78, 85, 72, 80, 90]
 
plt.plot(subjects, marks)
plt.title('Marks vs Subjects')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.show()
```

```python
plt.plot(subjects, marks)
plt.grid(axis = 'x')
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 100, 170, 160, 180]
 
plt.bar(months, sales)
plt.title('Monthly Sales')
plt.show()
```

```python
print("SCATTER PLOT")
study_hours = [1,2,3,4,5,6,7]
scores = [40,45,55,65,70,78,85]
 
plt.scatter(study_hours, scores)
plt.xlabel('Study Hours')
plt.ylabel('Scores')
plt.grid(True)
plt.show()
```

```python
days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
city1 = [30,32,31,33,34,35,36]
city2 = [25,26,27,26,28,29,30]
 
plt.plot(days, city1, label='City A')
plt.plot(days, city2, label='City B')
plt.legend()
plt.show()
```

```python
plt.subplot(2,1,1)
plt.plot(subjects, marks)
plt.subplot(2,1,2)
plt.bar(months, sales)
plt.show()
```

```python
obtained = [78,85,72,80,90]
 
plt.pie(obtained, labels=subjects, autopct='%1.1f%%')
# use explode=(0,0,0,0.1,0) to show coming out from chart
plt.title('Marks Distribution')
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr"] 
visitors = [1200, 1800, 2100, 1600] 
plt.plot(months, visitors, color="blue", marker="o", linestyle="--") 
plt.title("Monthly Website Visitors") 
plt.xlabel("Month") 
plt.ylabel("Visitors") 
plt.grid(True)
```

```python
products = ["Laptops", "Phones", "Tablets"] # Categories
sales = [200, 350, 150] # Values
plt.bar(products, sales, color=["#4CAF50", "#2196F3", "#FF5722"], width=0.6) 
plt.title("Q1 Product Sales")
plt.ylabel("Units Sold (Thousands)")
plt.ylim(0, 400) 
# to add text on each 
for i, v in enumerate(sales):
 plt.text(i, v, str(v), ha='center')
```
